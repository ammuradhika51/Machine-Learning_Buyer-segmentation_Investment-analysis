# Data Preprocessing
This notebook loads, cleans, and merges the datasets.

```
import pandas as pd

clients=pd.read_csv('clients.csv')
properties=pd.read_csv('properties.csv')

print(clients.head())
print(properties.head())
```

```
clients=clients.drop_duplicates()
properties=properties.drop_duplicates()

clients=clients.fillna(method='ffill')
properties=properties.fillna(method='ffill')

merged=clients.merge(properties,on='client_id',how='inner')
merged.to_csv('merged_dataset.csv',index=False)
merged.head()
```
